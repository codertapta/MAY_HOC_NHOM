#### nạp dữ liệu Marketing Campaign.

```python
import pandas as pd
from autoviz.AutoViz_Class import AutoViz_Class
 
# If the file is tab-separated (original Kaggle version), use sep='\t'
marketing_data = pd.read_csv("marketing_campaign.csv", sep='\t')

marketing_data.head()
```

####  chạy AutoViz và lưu các biểu đồ dạng HTML.

```python
AV = AutoViz_Class()
 
report = AV.AutoViz(
    filename = "",
    dfte = marketing_data,
    chart_format = "html",
    save_plot_dir = "Reports/autoviz"
)
```

#### chạy AutoViz với biến mục tiêu Response để xem quan hệ của các biến với Response.

```python
report_dep = AV.AutoViz(
    filename = "",
    dfte = marketing_data,
    depVar = "Response",
    chart_format = "svg",
    max_rows_analyzed = 150000,
    max_cols_analyzed = 30
)
```

#### Kết quả: AutoViz in ra các biểu đồ phân bố từng biến, scatter plot các cặp biến liên quan, heatmap tương quan và danh sách cảnh báo chất lượng dữ liệu (ví dụ cột Income có giá trị thiếu, các cột Z_CostContact/Z_Revenue là hằng số).
