#### Nạp dữ liệu và loại bỏ dòng thiếu giá trị.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
 
penguins_data = pd.read_csv("penguins_size.csv")
penguins_data = penguins_data.dropna(subset=['culmen_length_mm',
                                             'culmen_depth_mm',
                                             'flipper_length_mm',
                                             'body_mass_g'])
```

#### Biểu đồ boxplot

```python
plt.figure(figsize= (12,6))
ax = sns.boxplot(data = penguins_data, x= 'species', y= 'body_mass_g')
ax.set_xlabel('Penguin Species', fontsize = 15)
ax.set_ylabel('Body mass in g', fontsize = 15)
ax.set_title('Bivariate analysis of Species and body mass', fontsize= 20)
plt.show()
```

#### Violin plot (hiển thị thêm hình dạng phân bố mật độ của từng nhóm)

```python
plt.figure(figsize= (12,6))
ax = sns.violinplot(data = penguins_data, x= 'species', y= 'body_mass_g')
ax.set_xlabel('Penguin Species', fontsize = 15)
ax.set_ylabel('Body mass in g', fontsize = 15)
ax.set_title('Bivariate analysis of Species and body mass', fontsize= 20)
plt.show()
```
