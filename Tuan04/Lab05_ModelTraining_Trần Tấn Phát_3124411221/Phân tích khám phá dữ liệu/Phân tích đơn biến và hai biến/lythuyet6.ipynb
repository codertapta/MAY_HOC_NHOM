#### nạp dữ liệu và loại bỏ dòng thiếu giá trị.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
 
penguins_data = pd.read_csv("penguins_size.csv")
penguins_data = penguins_data.dropna(subset=['culmen_length_mm',
                                             'culmen_depth_mm',
                                             'flipper_length_mm',
                                             'body_mass_g'])
```

#### scatter plot thể hiện quan hệ giữa chiều dài mỏ và khối lượng cơ thể, tô màu theo loài.

```python
plt.figure(figsize= (12,6))
ax = sns.scatterplot(data = penguins_data,
                     x= 'culmen_length_mm',
                     y= 'body_mass_g',
                     hue= 'species')
ax.set_xlabel('Culmen Length in mm', fontsize = 15)
ax.set_ylabel('Body mass in g', fontsize = 15)
ax.set_title('Scatter plot of Culmen Length and body mass', fontsize= 20)
plt.show()
```

#### heatmap của ma trận hệ số tương quan Pearson giữa các biến số.

```python
num_cols = ['culmen_length_mm', 'culmen_depth_mm',
            'flipper_length_mm', 'body_mass_g']
corr_matrix = penguins_data[num_cols].corr()
corr_matrix
 
plt.figure(figsize= (10,8))
ax = sns.heatmap(corr_matrix, annot= True, fmt= '.2f',
                 cmap= 'coolwarm', vmin= -1, vmax= 1)
ax.set_title('Correlation heatmap of penguin measurements', fontsize= 18)
plt.show()
```

#### nhận xét flipper_length_mm và body_mass_g có tương quan dương rất mạnh; culmen_depth_mm có tương quan âm với flipper_length_mm và body_mass_g. Đây là ví dụ của nghịch lý Simpson: khi gộp ba loài thì tương quan âm, nhưng trong từng loài thì thường dương, vì vậy nên tô màu theo species (hue) để không kết luận sai.
