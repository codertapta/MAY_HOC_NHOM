#### nạp dữ liệu Marketing Campaign.

```python
import pandas as pd
import sweetviz as sv
 
marketing_data = pd.read_csv("marketing_campaign.csv" , sep='\t' )
marketing_data.head()

```

#### tạo báo cáo EDA tổng quát bằng analyze().

```python
report = sv.analyze(marketing_data)
report.show_html("Reports/sweetviz_report.html")
```

#### đặt biến mục tiêu (cột Response: khách hàng có chấp nhận chiến dịch cuối hay không).

```python
report_target = sv.analyze(marketing_data, target_feat= 'Response')
report_target.show_html("Reports/sweetviz_target.html")
```

#### so sánh hai nhóm khách hàng trong cùng bộ dữ liệu bằng compare_intra().

```python
report_compare = sv.compare_intra(marketing_data,
                                  marketing_data['Marital_Status'] == 'Married',
                                  ['Married', 'Not married'])
report_compare.show_html("Reports/sweetviz_compare.html")
```

#### Kết quả: mỗi lệnh show_html() sinh một file HTML và mở trên trình duyệt. Ưu điểm của SweetViz là so sánh hai tập/hai nhóm rất trực quan;
