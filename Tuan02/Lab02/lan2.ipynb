```python
import pandas as pd

columns = [
    'SalePrice',
    'GrLivArea',
    'LotArea',
    'TotalBsmtSF',
    '1stFlrSF',
    'GarageArea'
]

iqr_result = []

for col in columns:
    Q1 = train[col].quantile(0.25)
    Q3 = train[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = train[(train[col] < lower_bound) | (train[col] > upper_bound)]
    
    iqr_result.append([
        col,
        Q1,
        Q3,
        IQR,
        lower_bound,
        upper_bound,
        len(outliers)
    ])

iqr_df = pd.DataFrame(
    iqr_result,
    columns=[
        'Feature',
        'Q1',
        'Q3',
        'IQR',
        'Lower Bound',
        'Upper Bound',
        'Số lượng Outlier'
    ]
)

iqr_df
```

```python
from scipy.stats import zscore

columns = [
    'SalePrice',
    'GrLivArea',
    'LotArea',
    'TotalBsmtSF',
    '1stFlrSF',
    'GarageArea'
]

zscore_result = []

for col in columns:
    z = zscore(train[col])
    outliers = train[abs(z) > 3]
    
    zscore_result.append([
        col,
        len(outliers)
    ])

zscore_df = pd.DataFrame(
    zscore_result,
    columns=['Feature', 'Số lượng Outlier']
)

zscore_df
```
